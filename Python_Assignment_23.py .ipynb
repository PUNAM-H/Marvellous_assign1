```python
Q1. Write a Python program using multiprocessing.Pool to calculate the sum of all even numbers from 1 to N for every number from the given list.
```

```python
import multiprocessing
import os

def SumEven(N):
    Sum = 0

    for i in range(2, N + 1, 2):
        Sum = Sum + i

    print("Process ID :", os.getpid())
    print("Input Number :", N)
    print("Sum of Even Numbers :", Sum)
    print()

def main():

    Data = [1000000, 2000000, 3000000, 4000000]

    p = multiprocessing.Pool()

    p.map(SumEven, Data)

    p.close()
    p.join()

if __name__ == "__main__":
    main()
```

```python
Q5. Write a program that calculates factorials of multiple numbers simultaneously using multiprocessing.Pool.
```

```python
import multiprocessing
import os

def Factorial(N):

    Fact = 1

    for i in range(1, N + 1):
        Fact = Fact * i

    print("Process ID :", os.getpid())
    print("Input Number :", N)
    print("Factorial :", Fact)
    print()

def main():

    Data = [10, 15, 20, 25]

    p = multiprocessing.Pool()

    p.map(Factorial, Data)

    p.close()
    p.join()

if __name__ == "__main__":
    main()
```
