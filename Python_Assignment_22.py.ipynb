# Q1. Write a program that accepts a list of integers and uses Pool.map() to calculate the sum of squares from 1 to N for every element in the list.

```python
import multiprocessing

def SumSquare(N):
    Sum = 0

    for i in range(1, N + 1):
        Sum = Sum + (i * i)

    return Sum


def main():

    Data = [1000000, 2000000, 3000000, 4000000]

    pobj = multiprocessing.Pool()

    Result = pobj.map(SumSquare, Data)

    pobj.close()
    pobj.join()

    print("Sum of Squares : ")
    print(Result)


if __name__ == "__main__":
    main()
```

# Q2. Write a program that calculates factorials of multiple numbers simultaneously using Pool.map().

```python
import multiprocessing
import os

def Factorial(No):

    Fact = 1

    for i in range(1, No + 1):
        Fact = Fact * i

    return (os.getpid(), No, Fact)


def main():

    Data = [10, 15, 20, 25]

    pobj = multiprocessing.Pool()

    Result = pobj.map(Factorial, Data)

    pobj.close()
    pobj.join()

    print("Process ID\tInput\tFactorial")

    for Value in Result:
        print(Value[0], "\t", Value[1], "\t", Value[2])


if __name__ == "__main__":
    main()
```

# Q3. For every number in the given list, count how many prime numbers exist between 1 and N using multiprocessing Pool.

```python
import multiprocessing

def CountPrime(No):

    CountPrimeNo = 0

    for i in range(2, No + 1):

        Count = 0

        for j in range(1, i + 1):
            if i % j == 0:
                Count = Count + 1

        if Count == 2:
            CountPrimeNo = CountPrimeNo + 1

    return CountPrimeNo


def main():

    Data = [10000, 20000, 30000, 40000]

    pobj = multiprocessing.Pool()

    Result = pobj.map(CountPrime, Data)

    pobj.close()
    pobj.join()

    print("Prime Count : ")

    for Value in zip(Data, Result):
        print("Between 1 and", Value[0], "=", Value[1])


if __name__ == "__main__":
    main()
```

# Q4. Write a program that calculates
1⁵ + 2⁵ + 3⁵ + ... + N⁵

for multiple values of N simultaneously using Pool.map() and measure total execution time.

Input

```python
import multiprocessing
import time

def PowerSum(No):

    Sum = 0

    for i in range(1, No + 1):
        Sum = Sum + (i ** 5)

    return Sum


def main():

    Data = [10000, 20000, 30000, 40000]

    Start = time.time()

    pobj = multiprocessing.Pool()

    Result = pobj.map(PowerSum, Data)

    pobj.close()
    pobj.join()

    End = time.time()

    print("Result : ")

    for Value in zip(Data, Result):
        print("N =", Value[0], "Sum =", Value[1])

    print("Execution Time :", End - Start, "seconds")


if __name__ == "__main__":
    main()
```
