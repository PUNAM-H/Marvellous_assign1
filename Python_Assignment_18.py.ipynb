## 2) Write a program which accepts N numbers from the user and stores them into a List. Return the maximum (largest) number from that list.

# 3) Write a program which accepts N numbers from the user and stores them into a List. Return the minimum (smallest) number from that list.

```python
def main():
    element=int(input("Enter the list element number: "))
    num=[]

    for i in range (element):
        n=int(input("Enter the list elements number: "))
        num.append(n)
    
    min=num[0]
    for element in num:
        if element<min:
            min=element
    print("Minimum value is:",min)

if __name__ == "__main__":
    main()
```

# 4) Write a program which accepts N numbers from the user and stores them into a List. Accept one extra number from the user and return the frequency (count) of that number in the list.

```python
element=int(input("Enter the number:"))
data=[]
count=0
for i in range(element):
    num=int(input("Enter element:"))
    data.append(num)
    
search=int(input("Enter the number:"))
for num in data:
    if num == search:
        count=count+1
        
print("List:",data)
print("Frequency is:",count)
        
        
```

```python
def Frequency(arr, value):
    Count = 0

    for i in arr:
        if i == value:
            Count = Count + 1

    return Count


def main():
    size = int(input("Enter number of elements : "))

    Data = []

    print("Enter the elements : ")

    for i in range(size):
        no = int(input())
        Data.append(no)

    Search = int(input("Enter the element to search : "))

    Ans = Frequency(Data, Search)

    print("Frequency is :", Ans)


if __name__ == "__main__":
    main()
```
