# Q3. Accept one number and return its factorial.
5=1*2*3*4*5

```python
def Fact(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    print(f"Factorial of {n} is ",fact)
    
def main():
    number=int(input("Enter the number:"))
    Fact(number)
    
if __name__ == "__main__":
    main()
```

# Q4. Accept one number and return addition of its factors.

```python
def AddFactors(no):
    Factors=[]
    SumOfFactor=0
    
    for i in range(1,no):
        if no%i==0:
            Factors.append(i)
            SumOfFactor=SumOfFactor + i
            
    print("Factors:", Factors)        
    return SumOfFactor
    

def main():
     number=int(input("Enter the number:"))
     ans=AddFactors(number)
     print("Sum of Factor :" ,ans)
    
if __name__ == "__main__":
    main()
```

# Q5. Accept one number and check whether it is Prime or not.

```python
def CheckPrime(no):
    count = 0

    for i in range(1, no + 1):
        if no % i == 0:
            count = count + 1

    if count == 2:
        return True
    else:
        return False

def main():
    num = int(input("Enter number : "))

    if CheckPrime(num):
        print("Prime Number")
    else:
        print("Not Prime Number")

if __name__ == "__main__":
    main()
```

# Q9. Accept one number and return number of digits in that number.
