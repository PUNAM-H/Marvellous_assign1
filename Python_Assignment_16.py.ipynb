# Q3. Write a program which contains one function named Add() which accepts two numbers and returns their addition.

```python
def Add(no1,no2):
    return no1+no2
        
def main():
    num1 = int(input("Enter number: "))
    num2 = int(input("Enter number: "))
    ans=Add(num1,num2)
    print(ans)
   
if __name__ == "__main__":
    main()
```

# Q5. Write a program which displays numbers from 10 to 1.

```python
def Display(no):
    for i in range(no,0,-1):
        print(i)
        
def main():
    num = int(input("Enter number: "))
    Display(num)
   
if __name__ == "__main__":
    main()
```

# Q6. Write a program which accepts a number and checks whether it is Positive, Negative, or Zero.

```python
def Display(no):
    if no == 0:
        print("It's Zero")
    elif no>0:
        print("It's Positive")
    else:
        print("It's Negative")
        
def main():
    num = int(input("Enter number: "))
    Display(num)
    
if __name__ == "__main__":
    main()
```

# Q7. Write a program which accepts one number and returns True if it is divisible by 5, otherwise returns False.

```python
def Display(no):
    if no % 5 == 0:
        return True
    else:
        return False
    
def main():
    num = int(input("Enter number: "))
    ans=Display(num)
    print(ans)
    
if __name__ == "__main__":
    main()
```

# Q8. Write a program which accepts a number and prints that many * on the screen.

```python
def Display(no):
    for i in range(no):
        print("*",end=" ")

def main():
    num = int(input("Enter number: "))
    Display(num)

if __name__ == "__main__":
    main()
```

# Q9. Write a program which displays the first 10 even numbers.

```python
def DisplaysEven():
    
    Even=[]
    count=0
    
    for i in range(2,100,2):
        Even.append(i)
        count=count+1
        if count==10:
            break
            
    return Even  

def main():
    ans=DisplaysEven()
    print(ans)
    
if __name__ == "__main__":
    main()
```

# Q10. Write a program which accepts a name from the user and displays the length of that name.

```python
def Length(name):
    count=0
    for i in name:
        count=count+1
    return count
        
        
def main():
    Name=input("Enter the name:")
    Ans=Length(Name)
    print("Length of name is:", Ans)
    
if __name__ == "__main__":
    main()
```
