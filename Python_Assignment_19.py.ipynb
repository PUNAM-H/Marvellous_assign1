# Q1. Write a program which contains one lambda function which accepts one parameter and return power of two.

```python
def main():
    num = int(input("Enter number: "))
    
    power=lambda num:num*num
    
    result = power(num)
    
    print("The power of number:",result)

if __name__ == "__main__":
    main()
```

# Q3. Use filter(), map() and reduce()
Logic:
Filter → Numbers between 70 and 90
Map → Increase each number by 10
Reduce → Product of all numbers

```python
from functools import reduce
def main():
    user_list=[]
    
    size=int(input("Enter the size of list"))
    
    for i in range(size):
        no = int(input("Enter number: "))
        user_list.append(no)
    print("user list:",user_list)
        
    FilterData=list(filter(lambda no: no>=70 and no<=90 ,user_list))
    print("List after filter:", FilterData)
    
    mapdata=list(map(lambda no:no+10 ,FilterData))
    print("List after map:",mapdata)
    
    reducedata=reduce(lambda a,b:a*b ,mapdata)
    print("List after reduce",reducedata)
    
if __name__ == "__main__":
    main()
```

```python
from functools import reduce


def main():

    numbers = []

    size = int(input("Enter number of elements: "))

    for i in range(size):
        no = int(input("Enter number: "))
        numbers.append(no)


    FilterData = list(filter(lambda no: no >= 70 and no <= 90, numbers))

    print("List after filter:", FilterData)


    MapData = list(map(lambda no: no + 10, FilterData))

    print("List after map:", MapData)


    ReduceData = reduce(lambda a, b: a * b, MapData)

    print("Output of reduce:", ReduceData)


if __name__ == "__main__":
    main()
```

# Q4. Filter even numbers, calculate square and add all numbers using reduce.

```python
def main():
    
    user_list=[]
    
    size=int(input("Enter the size of list:"))
    
    for i in range(size):
        num=int(input("Enter the number in list:"))
        user_list.append(num)
    print("user list:",user_list)
        
    filterdata=list(filter(lambda no:no%2==0 ,user_list))
    print("List after filter:",filterdata)  
    
    mapdata=list(map(lambda no:no*no ,filterdata))
    print("List after map:",mapdata)
    
    reducedata=reduce(lambda a,b:a+b,mapdata)
    print("List after reduce:",reducedata)
    
if __name__ == "__main__":
    main()   
```

```python
from functools import reduce


def main():

    numbers = []

    size = int(input("Enter number of elements: "))

    for i in range(size):
        no = int(input("Enter number: "))
        numbers.append(no)


    EvenNumbers = list(filter(lambda no: no % 2 == 0, numbers))

    print("List after filter:", EvenNumbers)


    SquareNumbers = list(map(lambda no: no * no, EvenNumbers))

    print("List after map:", SquareNumbers)


    Addition = reduce(lambda a, b: a + b, SquareNumbers)

    print("Output of reduce:", Addition)


if __name__ == "__main__":
    main()
```

```python
from functools import reduce


def main():

    numbers = []

    size = int(input("Enter number of elements: "))

    for i in range(size):
        no = int(input("Enter number: "))
        numbers.append(no)


    PrimeNumbers = list(filter(lambda no: no > 1 and all(no % i != 0 for i in range(2, no)), numbers))

    print("List after filter:", PrimeNumbers)


    MultiplyNumbers = list(map(lambda no: no * 2, PrimeNumbers))

    print("List after map:", MultiplyNumbers)


    Maximum = reduce(lambda a, b: a if a > b else b, MultiplyNumbers)

    print("Output of reduce:", Maximum)


if __name__ == "__main__":
    main()
```
