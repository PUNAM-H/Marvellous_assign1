```python
import threading


def EvenFactor(number):

    print("Even Factors:")

    sum_even = 0

    for i in range(1, number + 1):

        if number % i == 0 and i % 2 == 0:
            print(i)
            sum_even = sum_even + i

    print("Sum of Even Factors:", sum_even)



def OddFactor(number):

    print("Odd Factors:")

    sum_odd = 0

    for i in range(1, number + 1):

        if number % i == 0 and i % 2 != 0:
            print(i)
            sum_odd = sum_odd + i

    print("Sum of Odd Factors:", sum_odd)



def main():

    num = int(input("Enter number: "))


    even_thread = threading.Thread(
        target=EvenFactor,
        args=(num,),
        name="EvenFactor"
    )


    odd_thread = threading.Thread(
        target=OddFactor,
        args=(num,),
        name="OddFactor"
    )


    even_thread.start()

    odd_thread.start()


    even_thread.join()

    odd_thread.join()


    print("Exit from main")



if __name__ == "__main__":
    main()
```

```python
3. Question:

Design a Python application that creates two threads named EvenList and OddList.

Both threads should accept a list of integers as input.
The EvenList thread should:
Extract all even elements from the list.
Calculate and display their sum.
The OddList thread should:
Extract all odd elements from the list.
Calculate and display their sum.
Threads should run concurrently.
```

```python
import threading


def EvenList(list1):

    print("Even Elements:")

    sum_even = 0

    for i in list1:

        if i % 2 == 0:
            print(i)
            sum_even = sum_even + i

    print("Sum of Even Elements:", sum_even)



def OddList(list1):

    print("Odd Elements:")

    sum_odd = 0

    for i in list1:

        if i % 2 != 0:
            print(i)
            sum_odd = sum_odd + i

    print("Sum of Odd Elements:", sum_odd)



def main():

    list1 = []

    size = int(input("Enter number of elements: "))


    for i in range(size):

        num = int(input("Enter element: "))
        list1.append(num)


    even_thread = threading.Thread(
        target=EvenList,
        args=(list1,),
        name="EvenList"
    )


    odd_thread = threading.Thread(
        target=OddList,
        args=(list1,),
        name="OddList"
    )


    even_thread.start()

    odd_thread.start()


    even_thread.join()

    odd_thread.join()


    print("Exit from main")


if __name__ == "__main__":
    main()
```

```python
4. Question:

Design a Python application that creates three threads named Small, Capital, and Digits.

All threads should accept a string as input.

Small thread should count lowercase characters.
Capital thread should count uppercase characters.
Digits thread should count numeric digits.

Each thread must display:

Thread ID
Thread Name
```

```python
import threading

def small(string):
    print("Thread1:")
    
    small_char_list=[]
    count=0
    
    for char in string:
        if char.islower():
            small_char_list.append(char)
            count=count+1
            
    print("The list of small laters: ",small_char_list)                  
    print("The count of small laters: ",count)
            
            
    
def capital(string):
    
    print("Thread2:")
    
    capital_char_list=[]
    count = 0
    
    for char in string:
        if char.isupper():
            capital_char_list.append(char)
            count = count + 1 
            
    print("The list of capital laters: ",capital_char_list)                  
    print("The count of capital laters: ",count)
    
    
def digit(string):
    print("Thread3:")
    
    digit_list=[]
    count = 0
    for char in string:
        if char.isdigit():
            digit_list.append(char)
            count = count + 1
            
    print("The list of digit : ",digit_list)                  
    print("The count of digit : ",count)
            
            
    
    
def main():
    
    string=input("Enter the string:")
    
    thread1=threading.Thread(target=small,args=(string,),name="small")
    thread2=threading.Thread(target=capital,args=(string,),name="capital")
    thread3=threading.Thread(target=digit,args=(string,),name="digit")
    
    thread1.start()
    thread2.start()
    thread3.start()
    
    thread1.join()
    thread2.join()
    thread3.join()
    
    
if __name__ == "__main__":
    main()
```

```python
import threading


def Small(string):

    count = 0

    for ch in string:

        if ch.islower():
            count += 1


    print("Thread Name:", threading.current_thread().name)
    print("Thread ID:", threading.get_ident())
    print("Lowercase Characters:", count)



def Capital(string):

    count = 0

    for ch in string:

        if ch.isupper():
            count += 1


    print("Thread Name:", threading.current_thread().name)
    print("Thread ID:", threading.get_ident())
    print("Uppercase Characters:", count)



def Digits(string):

    count = 0

    for ch in string:

        if ch.isdigit():
            count += 1


    print("Thread Name:", threading.current_thread().name)
    print("Thread ID:", threading.get_ident())
    print("Digits:", count)



def main():

    string = input("Enter string: ")


    small_thread = threading.Thread(
        target=Small,
        args=(string,),
        name="Small"
    )


    capital_thread = threading.Thread(
        target=Capital,
        args=(string,),
        name="Capital"
    )


    digit_thread = threading.Thread(
        target=Digits,
        args=(string,),
        name="Digits"
    )


    small_thread.start()

    capital_thread.start()

    digit_thread.start()


    small_thread.join()

    capital_thread.join()

    digit_thread.join()


    print("Exit from main")



if __name__ == "__main__":
    main()
```

# 5. Question:

Design a Python application that creates two threads named Thread1 and Thread2.

Thread1 should display numbers from 1 to 50.
Thread2 should display numbers from 50 to 1 in reverse order.
Thread2 starts execution only after Thread1 has completed.
Use appropriate thread synchronization.

```python
import threading


def display_forward():
    print("Thread1:")
    
    for i in range(1,50+1):
        print(i)
    
def display_reverse():
    print("Thread2:")
    
    for i in range(50,0,-1):
        print(i)
   


def main():
    
    Thread1=threading.Thread(target=display_forward ,name="display_forward")
    Thread2=threading.Thread(target=display_reverse ,name="display_reverse")

    Thread1.start()
    Thread2.start()
    
    Thread1.join()
    Thread2.join()

    print("Done.......")


if __name__ == "__main__":
    main()
```

```python
import threading


def Thread1():

    print("Thread1:")

    for i in range(1, 51):

        print(i)



def Thread2():

    print("Thread2:")

    for i in range(50, 0, -1):

        print(i)



def main():

    thread1 = threading.Thread(
        target=Thread1,
        name="Thread1"
    )


    thread2 = threading.Thread(
        target=Thread2,
        name="Thread2"
    )


    thread1.start()


    thread1.join()


    thread2.start()


    thread2.join()


    print("Exit from main")



if __name__ == "__main__":
    main()
```
